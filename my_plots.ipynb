```python
# Metric: reward minimal performance drop when decreasing retain_percent (start from 100%)
# Traverse from high x (100%) to low x; M = (1/N)*sum_i (Perf_{i-1} - Perf_i)*(x_{i-1} - x_i).
# Lower M = smaller weighted drop = better (strategy keeps performance when we reduce retention).
# Handles None by using only valid (x, y) pairs, then sorting by x descending.
import numpy as np

def weighted_marginal_gain(x_vals, y_vals):
    """
    x_vals, y_vals: lists (same length). y_vals may contain None.
    Traverses from 100% (max x) down. Returns (1/N)*sum(drop_i * step_i); lower = less drop = better.
    """
    n = min(len(x_vals), len(y_vals))
    x_plot = [x_vals[i] for i in range(n) if y_vals[i] is not None]
    y_plot = [y_vals[i] for i in range(n) if y_vals[i] is not None]
    if len(x_plot) < 2:
        return None
    # Sort by x descending: start from 100% (max retention), go down
    pairs = sorted(zip(x_plot, y_plot), key=lambda p: p[0], reverse=True)
    x_plot = [p[0] for p in pairs]
    y_plot = [p[1] for p in pairs]
    total_drop = 0.0
    for i in range(1, len(x_plot)):
        drop = y_plot[i - 1] - y_plot[i]  # performance drop when decreasing retention
        step = x_plot[i - 1] - x_plot[i]  # step in x (positive when going down)
        total_drop += drop * step
    N = len(x_plot) - 1
    # Lower = less drop = better
    return total_drop / N

# Same structure as combined plot: strategies with their x-axis dicts
dataset_mask_thresholds_per_ds = {
    "BigEarthNet": BigEarthNet_mask_thresholds,
    "MLRSNet": MLRSNet_mask_thresholds,
    "Woody": Woody_mask_thresholds,
    "Waititu": Waititu_mask_thresholds,
    "Flair": Flair_mask_thresholds,
    "MARIDA": Marida_mask_thresholds,
}
strategy_rows = [
    ("Stochastic", datasets_random_topx, {d: retain_percents for d in datasets_random_topx}),
    ("Diversity-based", datasets_topx, {d: retain_percents for d in datasets_topx}),
    ("Adaptive", datasets_max_len, dataset_mask_thresholds_per_ds),
]
models = ["vit_large", "vit_small", "vit_base"]

results = []
for strategy_name, datasets_dict, x_dict in strategy_rows:
    for dataset_name, model_data in datasets_dict.items():
        x_vals = x_dict[dataset_name]
        for model in models:
            values = list(model_data[model]["values"])
            m = weighted_marginal_gain(x_vals, values)
            results.append({"strategy": strategy_name, "dataset": dataset_name, "model": model, "metric": m})

# Print table: rows = (strategy, dataset), columns = model
import pandas as pd
df = pd.DataFrame(results)
pivot = df.pivot_table(index=["strategy", "dataset"], columns="model", values="metric")
print("Minimal-drop metric (from 100% down): (1/N)*sum(drop*dX). Lower = less performance drop.")
print(pivot.round(5).to_string())
pivot
```

```python
# Compare Stochastic vs Diversity-based per dataset and model (lower metric = better)
p_compare = pivot.loc[["Stochastic", "Diversity-based"]]
rows_list = []
for dataset in p_compare.index.get_level_values("dataset").unique():
    for model in p_compare.columns:
        s_val = p_compare.loc[("Stochastic", dataset), model]
        d_val = p_compare.loc[("Diversity-based", dataset), model]
        if pd.isna(s_val) and pd.isna(d_val):
            winner = "—"
            diff = None
        elif pd.isna(s_val):
            winner = "Diversity-based"
            diff = None
        elif pd.isna(d_val):
            winner = "Stochastic"
            diff = None
        elif s_val < d_val:
            winner = "Stochastic"
            diff = s_val - d_val
        elif d_val < s_val:
            winner = "Diversity-based"
            diff = s_val - d_val
        else:
            winner = "Tie"
            diff = 0.0
        rows_list.append({"dataset": dataset, "model": model, "Stochastic": s_val, "Diversity-based": d_val, "Winner": winner, "Diff (S−D)": diff})
compare_df = pd.DataFrame(rows_list)
winner_table = compare_df.pivot(index="dataset", columns="model", values="Winner")
print("Winner per dataset x model (lower metric = better):")
print(winner_table.to_string())
print("\nDifference (Stochastic - Diversity-based); negative = Stochastic better:")
diff_table = compare_df.pivot(index="dataset", columns="model", values="Diff (S−D)")
print(diff_table.round(5).to_string())
winner_table
```

```python
import os
import random
import cv2
import matplotlib.pyplot as plt

def plot_random_images_from_folders(high_unc_folder, low_unc_folder, num_samples=16, dataset = "Woody"):
    """
    Plots 10 random high and low uncertainty images side by side for each dataset.
    5 images will be on the first row and 5 on the second row.

    Args:
        high_unc_folder (str): Path to the folder containing high-uncertainty images.
        low_unc_folder (str): Path to the folder containing low-uncertainty images.
        num_samples (int): Number of random samples to select from each folder.
    """
    # Get all image files in the provided folders
    if dataset == "Woody":      
        dataset_name = "woody"
    elif dataset == "MARIDA":
        dataset_name = "marida"
    elif dataset == "BigEarthNet":
        dataset_name = "bigearthnet"
    elif dataset == "MLRSNet":  
        dataset_name = "MLRSNet"
    elif dataset == "Flair":
        dataset_name = "flair"
    elif dataset == "Waititu":
        dataset_name = "waititu"

    high_unc_folder = high_unc_folder + f"/{dataset_name}"
    low_unc_folder = low_unc_folder + f"/{dataset_name}"
    high_unc_images = [f for f in os.listdir(high_unc_folder) if f.endswith('.png') or f.endswith('.jpg')]
    low_unc_images = [f for f in os.listdir(low_unc_folder) if f.endswith('.png') or f.endswith('.jpg')]

    # Ensure there are enough images in both folders
    if len(high_unc_images) < num_samples or len(low_unc_images) < num_samples:
        raise ValueError("Not enough images in one of the folders.")

    # Randomly select `num_samples` images from both folders
    selected_high_unc_images = random.sample(high_unc_images, num_samples)
    selected_low_unc_images = random.sample(low_unc_images, num_samples)

    # Create the plot with 2 rows and 5 columns
    fig, axes = plt.subplots(2, 8, figsize=(6, 2))  # Adjusted for 5 images per row

    low_unc = []

    for i in range(8):
        # Read and convert images (OpenCV loads in BGR format)
        high_img = cv2.imread(os.path.join(high_unc_folder, selected_high_unc_images[i]))
        low_img = cv2.imread(os.path.join(low_unc_folder, selected_low_unc_images[i]))

        if dataset == "Waititu":
            while(low_img.max()) == 3 and selected_low_unc_images[i] in low_unc_images:
                selected_low_unc_images[i] = random.sample(low_unc_images, 1)[0]
                low_img = cv2.imread(os.path.join(low_unc_folder, selected_low_unc_images[i]))

        low_unc.append(selected_low_unc_images[i])

        if high_img is not None and low_img is not None:
            high_img = cv2.cvtColor(high_img, cv2.COLOR_BGR2RGB).astype(float)
            low_img = cv2.cvtColor(low_img, cv2.COLOR_BGR2RGB).astype(float)

            # Normalize images for display
            high_img /= high_img.max() if high_img.max() > 0 else 1
            low_img /= low_img.max() if low_img.max() > 0 else 1

            # Plot high-uncertainty image
            axes[0, i].imshow(high_img)
            axes[0, i].set_xticks([])  # Hide x-axis ticks
            axes[0, i].set_yticks([])  # Hide y-axis ticks
            # Plot low-uncertainty image
            axes[1, i].imshow(low_img)
            axes[1, i].set_xticks([])  # Hide x-axis ticks
            axes[1, i].set_yticks([]) 

    # Add row labels for clarity
    axes[0, 0].set_ylabel("High", fontsize=6, rotation=90, labelpad=10, x=0.8, y=0.5)
    axes[1, 0].set_ylabel("Low", fontsize=6, rotation=90, labelpad=10)

    plt.suptitle(f"{dataset}", fontsize=8, x=0.52, y=0.82)  # Add title

    
    # Adjust layout to reduce whitespace
    plt.subplots_adjust(hspace=0., wspace=0.)
      # Adjust gaps between images
    plt.tight_layout()
    fig.savefig(f"paper_images/{dataset}.png", dpi=300, bbox_inches='tight', pad_inches=0.0)

    # Show the plot
    plt.show()
high_path = "/mnt/shared_storage/nbountos/Projects/Uncertain-Eval-Dev/overleaf_uncertainty_samples_plots/high_uncertainty"
low_path = "/mnt/shared_storage/nbountos/Projects/Uncertain-Eval-Dev/overleaf_uncertainty_samples_plots/low_uncertainty"
# Example usage:
plot_random_images_from_folders(high_path,low_path, dataset="Flair")
```
